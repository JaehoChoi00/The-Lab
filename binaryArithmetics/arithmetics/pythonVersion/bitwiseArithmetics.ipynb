```python
def binaryAnd(bitA, bitB):
    if (bitA == 1 and bitB == 1):
        return 1
    return 0

def binaryOR(bitA, bitB):
    if (bitA == 1 or bitB == 1):
        return 1
    return 0

def binaryXOR(bitA, bitB):
    if (bitA == bitB):
        return 0
    return 1

def binaryNOT(bit):
    return 1 if bit == 0 else 0

def shiftLeft(binary):
    temp = ""
    for bit in range(1, len(binary)):
        temp = temp + binary[bit]
    temp = temp + "0"
    return temp

def flipBinary(binary):
    flippedBinary = ""
    for index in range (len(binary)):
        flippedBinary = flippedBinary + str(binaryNOT(int(binary[index])))
    return flippedBinary

def binaryToVariable(binary):
    conversionGrid = [2 ** i for i in range (len(binary) - 1, -1, -1)]
    variable = 0
    for index in range(len(binary) - 1, -1, -1):
        if (int(binary[index]) == 1):
            variable = variable + int(conversionGrid[index])
    return variable
```

```python
def alu_add(binaryA, binaryB):
    if int(binaryB) == 0:
        return binaryA

    while (int(binaryB) != 0):
        carryAnd = ""
        temporarySum = ""
        for index in range(len(binaryA) - 1, -1, -1):
            carryAnd = str(binaryAnd(int(binaryA[index]), int(binaryB[index]))) + carryAnd
            temporarySum = str(binaryXOR(int(binaryA[index]), int(binaryB[index]))) + temporarySum

        binaryA = temporarySum
        binaryB = shiftLeft(carryAnd)
        
    return binaryA


print(alu_add("00101101", "00011011"))
```

```python

def alu_sub(binaryA, binaryB):

    invertedBinaryB = flipBinary(binaryB)
    onePadded = "0" * (len(binaryB) - 1) + "1"

    twosCompliment = alu_add(invertedBinaryB, onePadded)

    return alu_add(binaryA, twosCompliment)

print(str(binaryToVariable("00101101")) + " " + str(binaryToVariable("00011011")))
print(binaryToVariable(alu_sub("00101101", "00011011")))
```
